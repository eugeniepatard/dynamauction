## Constant bid sweep

Grid search over the fixed bid value b for `ConstantPolicy` (η = 50).

#### Results for $\eta = 50$

```python
# Load sigmoid constant-bid sweep CSVs (η=50)
# Gross = wide bid range (max bid > 2.0), refined = narrow range (max bid <= 2.0)
_pattern_sig_cb = "benchmark_results/sigmoid_results/constant-bid/survival-second-price-exponential_eta50_sqsigmoid_constant_bid_sweep*.csv"
_files_sig_cb = sorted(glob.glob(_pattern_sig_cb))
if not _files_sig_cb:
    raise FileNotFoundError(
        f"No sigmoid constant-bid sweep CSV found matching {_pattern_sig_cb!r}.\n"
        "Run:  python benchmark/tune_constant_bid.py --horizon-eta 50 --squashing sigmoid"
    )

_dfs_sig_cb = [(f, pd.read_csv(f)) for f in _files_sig_cb]
_gross_sig  = [(f, df) for f, df in _dfs_sig_cb if df["bid_value"].max() > 2.0]
_refined_sig = [(f, df) for f, df in _dfs_sig_cb if df["bid_value"].max() <= 2.0]

if not _gross_sig:
    raise FileNotFoundError("No sigmoid gross constant-bid CSV found (expected max bid > 2.0).")
if not _refined_sig:
    raise FileNotFoundError("No sigmoid refined constant-bid CSV found (expected max bid <= 2.0).")

constant_sig_gross_df = _gross_sig[-1][1]
constant_sig_refined_df = _refined_sig[-1][1]
print(f"Sigmoid gross:   {_gross_sig[-1][0]}")
print(f"Sigmoid refined: {_refined_sig[-1][0]}")

constant_sig_gross_summary = (
    constant_sig_gross_df.sort_values(["bid_value", "seed", "step"])
    .groupby(["bid_value", "seed"])["cum_reward"].last()
    .reset_index()
    .rename(columns={"cum_reward": "episode_reward"})
)
constant_sig_refined_summary = (
    constant_sig_refined_df.sort_values(["bid_value", "seed", "step"])
    .groupby(["bid_value", "seed"])["cum_reward"].last()
    .reset_index()
    .rename(columns={"cum_reward": "episode_reward"})
)
```

```python
mean_r = constant_sig_gross_summary.groupby("bid_value")["episode_reward"].mean()
std_r  = constant_sig_gross_summary.groupby("bid_value")["episode_reward"].std()
best_bid_sig_gross = mean_r.idxmax()
best_reward_sig_gross = mean_r.loc[best_bid_sig_gross]

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(mean_r.index, mean_r.values, marker="o", lw=2)
ax.scatter(best_bid_sig_gross, best_reward_sig_gross, color="red", zorder=5)
ax.fill_between(mean_r.index, mean_r - std_r, mean_r + std_r, alpha=0.2, label="±1 std")
ax.axhline(best_reward_sig_gross, color="green", linestyle="--")
ax.axvline(best_bid_sig_gross, color="red", linestyle="--", label=f"Best b = {best_bid_sig_gross:.4g}")
ax.text(ax.get_xlim()[0], best_reward_sig_gross, f"{best_reward_sig_gross:.3f}",
        va="bottom", ha="right", color="green", fontsize=8)
ax.set_xlabel("Constant bid value b")
ax.set_ylabel("Mean episode reward")
ax.set_title("Constant bid sweep gross scale — mean episode reward (sigmoid, η = 50)")
ax.legend()
plt.tight_layout()
os.makedirs("plots", exist_ok=True)
plt.savefig("plots/sigmoid_constant_bid_sweep_gross_mean_eta50.pdf", bbox_inches="tight")
plt.show()
print(f"Best b = {best_bid_sig_gross:.4g}  (mean reward = {best_reward_sig_gross:.4f})")
```

```python
mean_r = constant_sig_refined_summary.groupby("bid_value")["episode_reward"].mean()
std_r  = constant_sig_refined_summary.groupby("bid_value")["episode_reward"].std()
best_bid_sig_refined = mean_r.idxmax()
best_reward_sig_refined = mean_r.loc[best_bid_sig_refined]

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(mean_r.index, mean_r.values, marker="o", lw=2)
ax.scatter(best_bid_sig_refined, best_reward_sig_refined, color="red", zorder=5)
ax.fill_between(mean_r.index, mean_r - std_r, mean_r + std_r, alpha=0.2, label="±1 std")
ax.axhline(best_reward_sig_refined, color="green", linestyle="--")
ax.axvline(best_bid_sig_refined, color="red", linestyle="--", label=f"Best b = {best_bid_sig_refined:.4g}")
ax.text(ax.get_xlim()[0], best_reward_sig_refined, f"{best_reward_sig_refined:.3f}",
        va="bottom", ha="left", color="green", fontsize=8)
ax.set_xlabel("Constant bid value b")
ax.set_ylabel("Mean episode reward")
ax.set_title("Constant bid sweep refined scale — mean episode reward (sigmoid, η = 50)")
ax.legend()
plt.tight_layout()
plt.savefig("plots/sigmoid_constant_bid_sweep_refined_mean_eta50.pdf", bbox_inches="tight")
plt.show()
print(f"Best b = {best_bid_sig_refined:.4g}  (mean reward = {best_reward_sig_refined:.4f})")
```

```python
fig, ax = plt.subplots(figsize=(10, 4))
bids_sorted_sig = sorted(constant_sig_refined_summary["bid_value"].unique())
sns.boxplot(
    data=constant_sig_refined_summary, x="bid_value", y="episode_reward",
    order=bids_sorted_sig, ax=ax,
)
ax.set_xticklabels([f"{b:.2f}" for b in bids_sorted_sig], rotation=45, ha="right")
ax.axvline(bids_sorted_sig.index(best_bid_sig_refined), color="red", linestyle="--",
           label=f"Best b = {best_bid_sig_refined:.4g}")
ax.set_xlabel("Constant bid value b")
ax.set_ylabel("Episode reward")
ax.set_title("Constant bid sweep refined scale — episode reward distribution (sigmoid, η = 50)")
ax.legend()
plt.tight_layout()
plt.savefig("plots/sigmoid_constant_bid_sweep_dist_eta50.pdf", bbox_inches="tight")
plt.show()
```

## Shading α sweep

Grid search over the shading multiplier α for `ShadingPolicy` (η = 50).

#### Results for $\eta = 50$

```python
# Load sigmoid shading alpha sweep CSV (η=50) — use merged file if available
_pattern_sig_alpha = "benchmark_results/sigmoid_results/alpha_shading/survival-second-price-exponential_eta50_sqsigmoid_shading_alpha_sweep_merged.csv"
_files_sig_alpha = sorted(glob.glob(_pattern_sig_alpha))
if not _files_sig_alpha:
    # Fall back to any individual run file
    _pattern_sig_alpha = "benchmark_results/sigmoid_results/alpha_shading/survival-second-price-exponential_eta50_sqsigmoid_shading_alpha_sweep*.csv"
    _files_sig_alpha = sorted(glob.glob(_pattern_sig_alpha))
if not _files_sig_alpha:
    raise FileNotFoundError(
        "No sigmoid shading alpha sweep CSV found.\n"
        "Run:  python benchmark/tune_shading_alpha.py --horizon-eta 50 --squashing sigmoid"
    )

shading_sig_df = pd.read_csv(_files_sig_alpha[-1])
print(f"Loaded: {_files_sig_alpha[-1]}")
print(f"Alphas: {sorted(shading_sig_df['alpha'].unique().round(4).tolist())}")

shading_sig_summary = (
    shading_sig_df.sort_values(["alpha", "seed", "step"])
    .groupby(["alpha", "seed"])["cum_reward"].last()
    .reset_index()
    .rename(columns={"cum_reward": "episode_reward"})
)
```

### Aggregate performance: n_rollouts comparison (sigmoid, η=50)

Three panels: **baseline** (n_rollouts=10), **partial update** (n_rollouts=30 — GreedyMyopic, MPC(H=10), MPC(H=25), MPCTimeBinned(H=10,K=2)), and **partial update** (n_rollouts=150 — GreedyMyopic, Shading(α=1) completed so far). Coloured bars indicate which algorithms were re-evaluated at each level.

```python
# Reload original n_rollouts=10 sigmoid data (unpatch)
_df_r10_orig = pd.read_csv(
    'benchmark_results/sigmoid_results/survival-second-price-exponential_eta50_sigmoid_results_final.csv'
)

def make_scores(df):
    ep = df.groupby(['policy', 'seed'])['reward'].sum().reset_index()
    ep.columns = ['policy', 'seed', 'episode_reward']
    return ep

# r10: baseline
scores_r10 = make_scores(_df_r10_orig)

# r30: df_eta50_sigmoid has GreedyMyopic, MPC(H=10), MPC(H=25), MPCTimeBinned(H=10,K=2) patched
scores_r30 = make_scores(df_eta50_sigmoid)
_r30_updated = {'GreedyMyopic', 'MPC(H=10)', 'MPC(H=25)', 'MPCTimeBinned(H=10,K=2)'}

# r150: patch GreedyMyopic and Shading(α=1) from the r150 CSVs
_r150_greedy  = pd.read_csv('benchmark_results/rollout150_results/sigmoid_eta50_greedy_r150.csv')
_r150_shading = pd.read_csv('benchmark_results/rollout150_results/sigmoid_eta50_shading_r150.csv')
_r150_updated = set(_r150_greedy['policy'].unique()) | set(_r150_shading['policy'].unique())
_df_r150 = pd.concat([
    df_eta50_sigmoid[~df_eta50_sigmoid['policy'].isin(_r150_updated)],
    _r150_greedy,
    _r150_shading,
], ignore_index=True)
scores_r150 = make_scores(_df_r150)

# Colour palette
COL_BASE = '#5b8db8'   # blue  — unchanged from previous panel
COL_R30  = '#f28e2b'   # orange — re-evaluated at n_rollouts=30
COL_R150 = '#59a14f'   # green  — re-evaluated at n_rollouts=150

def bar_colors(stats, updated, col_updated):
    return [col_updated if p in updated else COL_BASE for p in stats.index]

fig, axes = plt.subplots(1, 3, figsize=(22, 6), sharey=True)

for ax, scores, label, updated, col_upd in [
    (axes[0], scores_r10,  'n_rollouts=10 (baseline)',        set(),         COL_BASE),
    (axes[1], scores_r30,  'n_rollouts=30 (partial update)',  _r30_updated,  COL_R30),
    (axes[2], scores_r150, 'n_rollouts=150 (partial update)', _r150_updated, COL_R150),
]:
    stats = (
        scores.groupby('policy')['episode_reward']
        .agg(mean='mean', sem=lambda x: x.std() / np.sqrt(len(x)))
        .sort_values('mean', ascending=True)
    )
    colors = bar_colors(stats, updated, col_upd)
    ax.barh(stats.index, stats['mean'], xerr=stats['sem'],
            capsize=3, color=colors, alpha=0.85, height=0.6)
    ax.set_xlabel('Mean episode reward')
    ax.set_title(label, fontweight='bold')
    ax.grid(axis='x', alpha=0.3)

from matplotlib.patches import Patch
legend_elements = [
    Patch(facecolor=COL_BASE, alpha=0.85, label='unchanged (n_rollouts=10)'),
    Patch(facecolor=COL_R30,  alpha=0.85, label='re-evaluated at n_rollouts=30'),
    Patch(facecolor=COL_R150, alpha=0.85, label='re-evaluated at n_rollouts=150'),
]
fig.legend(handles=legend_elements, loc='lower center', ncol=3, bbox_to_anchor=(0.5, -0.06),
           fontsize=11)
fig.suptitle('Aggregate performance — sigmoid squashing (η=50)', fontweight='bold', fontsize=13)
fig.tight_layout()
plt.savefig('plots/aggregate_metrics_r10_r30_r150_sigmoid_eta50.pdf', bbox_inches='tight')
plt.show()
```

## n_rollouts=10 vs n_rollouts=30 comparison (sigmoid, η=50)

Re-evaluation of GreedyMyopic, MPC(H=10), MPC(H=25) and MPCTimeBinned(H=10,K=2) with 3× more MC rollouts per planning step (30 vs 10). MPCTimeBinned(H=10,K=3), (H=25,K=2) and (H=25,K=3) are still running.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Load rollout-30 partial results
df_r30 = pd.read_csv('benchmark_results/rollout30_results/sigmoid_eta50_rollout30_partial.csv')

def ep_rewards(df):
    return df.groupby(['policy','seed'])['reward'].sum().reset_index()

ep10 = ep_rewards(df_eta50_sigmoid)
ep30 = ep_rewards(df_r30)

policies_r30 = df_r30['policy'].unique().tolist()

stats10 = (
    ep10[ep10['policy'].isin(policies_r30)]
    .groupby('policy')['reward']
    .agg(mean='mean', sem=lambda x: x.std()/np.sqrt(len(x)))
    .rename(columns={'mean':'mean_r10','sem':'sem_r10'})
)
stats30 = (
    ep30.groupby('policy')['reward']
    .agg(mean='mean', sem=lambda x: x.std()/np.sqrt(len(x)))
    .rename(columns={'mean':'mean_r30','sem':'sem_r30'})
)

cmp = stats10.join(stats30).sort_values('mean_r30', ascending=False)
cmp['delta'] = cmp['mean_r30'] - cmp['mean_r10']
cmp['delta_sem'] = np.sqrt(cmp['sem_r10']**2 + cmp['sem_r30']**2)
print(cmp[['mean_r10','sem_r10','mean_r30','sem_r30','delta','delta_sem']].round(4).to_string())
```

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

labels = cmp.index.tolist()
x = np.arange(len(labels))
w = 0.35

ax1.bar(x - w/2, cmp['mean_r10'], w, yerr=cmp['sem_r10'],
        capsize=4, label='n_rollouts=10', alpha=0.85, color='steelblue')
ax1.bar(x + w/2, cmp['mean_r30'], w, yerr=cmp['sem_r30'],
        capsize=4, label='n_rollouts=30', alpha=0.85, color='darkorange')
ax1.set_xticks(x)
ax1.set_xticklabels(labels, rotation=30, ha='right')
ax1.set_ylabel('Mean episode reward')
ax1.set_title('Absolute performance')
ax1.legend()

colors = ['green' if d > 0 else 'red' for d in cmp['delta']]
ax2.bar(x, cmp['delta'], color=colors, alpha=0.8)
ax2.errorbar(x, cmp['delta'], yerr=cmp['delta_sem'], fmt='none',
             color='black', capsize=4)
ax2.axhline(0, color='black', lw=0.8)
ax2.set_xticks(x)
ax2.set_xticklabels(labels, rotation=30, ha='right')
ax2.set_ylabel('Delta mean reward (r30 - r10)')
ax2.set_title('Gain from 3x more rollouts')

fig.suptitle('n_rollouts=10 vs 30 -- sigmoid squashing (eta=50)', fontweight='bold')
fig.tight_layout()
plt.savefig('plots/rollout10_vs_30_sigmoid_eta50.pdf', bbox_inches='tight')
plt.show()
```

## n_rollouts=150 results (sigmoid, η=50)

Re-evaluation with 5× more Monte Carlo rollouts per planning step.
**GreedyMyopic** and **Shading(α=1.0)** completed (1000 seeds each).
MPC variants, MPCTimeBinned and the shading α sweep are still running.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load r150 results (available so far)
_r150_greedy = pd.read_csv('benchmark_results/rollout150_results/sigmoid_eta50_greedy_r150.csv')
_r150_shading = pd.read_csv('benchmark_results/rollout150_results/sigmoid_eta50_shading_r150.csv')
df_r150 = pd.concat([_r150_greedy, _r150_shading], ignore_index=True)
print('Policies in r150 so far:')
print(df_r150.groupby('policy')['seed'].nunique().rename('n_seeds').to_string())
```

```python
# Compare episode rewards across n_rollout levels for Greedy and Shading
_r10_orig = pd.read_csv(
    'benchmark_results/sigmoid_results/survival-second-price-exponential_eta50_sigmoid_results_final.csv'
)
_r30_patch = pd.read_csv('benchmark_results/rollout30_results/sigmoid_eta50_rollout30_partial.csv')

def ep_rew(df, policy):
    sub = df[df['policy'] == policy]
    return sub.groupby('seed')['reward'].sum()

policies_to_compare = ['GreedyMyopic', 'Shading(α=1)']
# Shading in r10 was stored as 'Shading(α=0.89)'
r10_shading_label = [c for c in _r10_orig['policy'].unique() if 'Shading' in c]
print('Shading labels in r10:', r10_shading_label)

rows = []
for label, df, n_r in [('r10 (n=10)', _r10_orig, 10),
                        ('r30 (n=30)', _r30_patch, 30),
                        ('r150 (n=150)', df_r150, 150)]:
    for pol in df['policy'].unique():
        if 'Greedy' in pol or 'Shading' in pol:
            ep = df[df['policy'] == pol].groupby('seed')['reward'].sum()
            rows.append({'rollout': label, 'n_rollouts': n_r,
                         'policy': 'GreedyMyopic' if 'Greedy' in pol else 'Shading',
                         'mean': ep.mean(), 'sem': ep.sem()})
df_cmp = pd.DataFrame(rows)
print(df_cmp.to_string(index=False))
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5), sharey=False)

for ax, pol in zip(axes, ['GreedyMyopic', 'Shading']):
    sub = df_cmp[df_cmp['policy'] == pol].sort_values('n_rollouts')
    ax.errorbar(sub['n_rollouts'], sub['mean'], yerr=sub['sem'],
                marker='o', lw=2, capsize=5, color='steelblue')
    ax.set_xlabel('n_rollouts')
    ax.set_ylabel('Mean episode reward')
    ax.set_title(pol, fontweight='bold')
    ax.set_xscale('log')
    ax.set_xticks([10, 30, 150])
    ax.set_xticklabels(['10', '30', '150'])
    ax.grid(axis='y', alpha=0.3)

fig.suptitle('Convergence with n_rollouts — sigmoid squashing (η=50, 1000 seeds)',
             fontweight='bold')
fig.tight_layout()
plt.savefig('plots/rollout_convergence_greedy_shading_sigmoid_eta50.pdf', bbox_inches='tight')
plt.show()
```

```python
# Also compare directly against FiPla reference level
_fipla_r10 = _r10_orig[_r10_orig['policy'] == 'FiPla'].groupby('seed')['reward'].sum()
fipla_mean = _fipla_r10.mean()
fipla_sem = _fipla_r10.sem()

fig, ax = plt.subplots(figsize=(9, 5))
colors = {'r10 (n=10)': '#aec7e8', 'r30 (n=30)': '#6baed6', 'r150 (n=150)': '#2171b5'}
x_pos = {'GreedyMyopic': 0, 'Shading': 1}
width = 0.22
offsets = {'r10 (n=10)': -width, 'r30 (n=30)': 0, 'r150 (n=150)': width}

for rollout_label, grp in df_cmp.groupby('rollout'):
    for _, row in grp.iterrows():
        x = x_pos[row['policy']] + offsets[rollout_label]
        ax.bar(x, row['mean'], width*0.9, yerr=row['sem'],
               capsize=4, color=colors[rollout_label],
               label=rollout_label if row['policy'] == 'GreedyMyopic' else None,
               alpha=0.9)

ax.axhline(fipla_mean, color='red', linestyle='--', lw=1.5, label=f'FiPla (oracle) = {fipla_mean:.3f}')
ax.fill_between([-0.4, 1.4], fipla_mean - fipla_sem, fipla_mean + fipla_sem,
                color='red', alpha=0.1)
ax.set_xticks([0, 1])
ax.set_xticklabels(['GreedyMyopic', 'Shading(α=1.0)'], fontsize=12)
ax.set_ylabel('Mean episode reward')
ax.set_title('Effect of n_rollouts vs FiPla oracle — sigmoid (η=50)', fontweight='bold')
ax.legend()
ax.grid(axis='y', alpha=0.3)
fig.tight_layout()
plt.savefig('plots/rollout_convergence_vs_fipla_sigmoid_eta50.pdf', bbox_inches='tight')
plt.show()
```
